```python
#Exploration of data using visualization(make graphical summary)
#===============================================================
#what is data visualization?
#-> It is useful to gain insights and understand what happened in the past in a given context.It is also helpful for feature engineering 

#Matplotlib is a python 2D plotting library and the most wildly used for data visualization. It makes use numpy arrays.
#It provides an expensive sets of plotting API to create various plot .

#Seaborn : It is also a python data visualization library built on top of matplotlib. It provides a high level interface for drawing innovative and informative statistical charts.

#To display the plots on the jupyter notebook, we need to provide %matplotlib inline

import matplotlib.pyplot as plt
import seaborn as sn
%matplotlib inline
```

```python
#PLOTTING X & Y POINTS
#=====================
#the plot function is draw points(markers) in a diagram.
#Qs: Draw a line in a diagram from position(1,3) to position(8,10)
#Ans->
import matplotlib.pyplot as plt
import numpy as np
xpoints = np.array([1,8])
ypoints = np.array([3,10])
plt.plot(xpoints,ypoints)
plt.show()
```

```python
#Qs: Draw a line in a diagram from position(1,3)to(2,8),(6,1)to(8,10)
#Ans:
import matplotlib.pyplot as plt
import numpy as np
xpoints = np.array([1,2,6,8])
ypoints = np.array([3,8,1,10])
plt.plot(xpoints,ypoints)
plt.show()
```

```python
#PLOTTING WITHOUT LINE
#Qs: Draw 2 points in the diagram & mark each point with a star from(1,3) to (8,10)
#Ans:
import matplotlib.pyplot as plt
import numpy as np
xpoints = np.array([1,8])
ypoints = np.array([3,10])
plt.plot(xpoints,ypoints,'*')
plt.show()
```

```python
#DEFAULT x POINTS
#================
#If we do not specify the points on x axis , they will get the default values 0,2..etc. depending on the length of the y points.
#You can use the key word 'marker' to emphasize each point with specify marker
import matplotlib.pyplot as plt
import numpy as np
ypoints = np.array([3,8,-2,-5,10])
plt.plot(ypoints,marker='*')
plt.show()
```

```python
#QS: FORMAT STRINGS fmt
#===========================
#marker|line|color
import matplotlib.pyplot as plt
import numpy as np
ypoints = np.array([3,8,1,10,-9])
plt.plot(ypoints,'o-.b')
plt.show()
```
