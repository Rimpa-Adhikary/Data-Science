```python
#ploting x and y points
# the plot() function is use to draw points(markers) in a diagram
import matplotlib.pyplot as plt
import numpy as np
#draw a line in a diagram from position (1,3)to position(8,10):
xpoints=np.array([1,8])
ypoints=np.array([3,10])
plt.plot(xpoints,ypoints)
plt.show
```

```python
#default x points
#if we donot specify the point on the x-axis they will get the default values0,1,2 depending on the length of the y points
#you can use the keyword"marker" to emphasise each point with a sepcified marker
ypoints=np.array([3,8,1,10,-9])

plt.plot(ypoints,marker='d')
plt.show
```

```python
import pandas as pd
ipl_auction_df=pd.read_csv('IPL dataset1 (CSBS-2023).csv')
```

```python
#mean of grouped data
soldprice_by_age=ipl_auction_df.groupby('AGE')['SOLD PRICE'].mean().reset_index()
print(soldprice_by_age)
```

```python
import seaborn as sn
sn.barplot(x='AGE',y='SOLD PRICE',data=soldprice_by_age)
plt.title("Bar plot for average sold price versus age")
plt.xlabel('AGE')
plt.ylabel('SOLID PRICE')
plt.show()
# write interpretation:
#the figure show that the  sold price is higher for age category 1.
```

```python
#histogram::a histogram is a plot that show the frequency distribution of a set of continuous variables.it gives an insight into the underlaying distribution of the variable outlier ,skewness,etc.to draw a histogram we invoke the hist() method of matplotlib.
plt.hist(ipl_auction_df['SOLD PRICE'])
plt.title("HISTROGRAM for sold price")
plt.xlabel('SOLD PRICE')
plt.ylabel('FREQUENCY')
plt.show()
#interpretion
#RIGHT skewed
#the histogram shows that sold price is right skewed
#a distribution is right skew if it has tail on the right side of the distribution
#a distribution is left skew if it has tail on the left side of the distribution
#most players are auction at low price range of 250000and 500000,whereas players are paid 
#by default it creates 10 bins in the histogram.
```

```python
#histogram::a histogram is a plot that show the frequency distribution of a set of continuous variables.it gives an insight into the underlaying distribution of the variable outlier ,skewness,etc.to draw a histogram we invoke the hist() method of matplotlib.
plt.hist(ipl_auction_df['TEAM'])
plt.title("HISTROGRAM for TEAM")
plt.xlabel('TEAM')
plt.ylabel('FREQUENCY')
plt.show()
#interpretion
#RIGHT skewed
#the histogram shows that sold price is right skewed
#a distribution is right skew if it has tail on the right side of the distribution
#a distribution is left skew if it has tail on the left side of the distribution
```

```python
#by default it creates 10 bins in the histogram.
plt.hist(ipl_auction_df['SOLD PRICE'],bins=20)
plt.title("HISTROGRAM for sold price")
plt.xlabel('SOLD PRICE')
plt.ylabel('FREQUENCY')
plt.show()
# bins are called interval
```

```python
plt.hist(ipl_auction_df['TEAM'],bins=30)
plt.title("HISTROGRAM for TEAM")
plt.xlabel('TEAM')
plt.ylabel('FREQUENCY')
plt.show()
```
